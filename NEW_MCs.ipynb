```python
import pandas as pd
```

```python
metal = ['Sc', 'Ti', 'V','Cr','Mn','Fe','Co','Ni','Cu','Zn', \
         'Y','Zr','Nb','Mo','Ru','Rh','Pd','Ag','Cd','Hf',\
         'Ta','W','Re','Os','Ir','Pt','Au','La','Ce',\
         'Al','Ga','In','Sn','Pb','Bi']   # 35
# 1-10

nonmetal = ['O'] 
# 1-10

second = ['Mn','Fe','Co','Ni','Ru','Pd','Ag','Pt']

element_ratio = [0.05, 0.1, 0.2]
```

```python
def gcd(a, b):
    while b:
        a, b = b, a % b
    return a

result = [(1, 1)]
for i in range(1, 11):
    for j in range(i+1, 11):
        if gcd(i, j) == 1:
            result.append((i, j))  # 32
```

```python
compound_O = []
for nm in nonmetal:
    for m in metal:
        for index in result:
            if index[0]==1 and index[1]==1:
                com = m + nm
            elif index[0] == 1:
                com = m + nm + f'{index[1]}'
            elif index[1] == 1:
                com = m + f'{index[0]}' + nm
            else:
                com = m + f'{index[0]}' + nm + f'{index[1]}'
            compound_O.append(com)
# compound
# compound = pd.DataFrame(compound, columns=['compound'])
# compound.to_csv('NEW_TMCs.csv', index=False)
compound_O
```

```python
Bicompound_O = []
for nm in nonmetal:
    for m in metal:
        for se in second:
            if se == m:
                continue
            for index in result:
                for er in element_ratio:
                    se_value = index[0] * er
                    m_value = index[0] * (1 - er)
                    
                    def is_close_to_int(x):
                        return abs(x - round(x)) < 1e-6
                    
                    se_index = f"{se_value:.2f}".rstrip('0').rstrip('.') if not is_close_to_int(se_value) else ''
                    m_index = f"{m_value:.2f}".rstrip('0').rstrip('.') if not is_close_to_int(m_value) else ''
                    nm_index = str(index[1]) if index[1] != 1 else ''
                    
                    com = f"{se}{se_index}{m}{m_index}{nm}{nm_index}"
                    if com not in Bicompound_O:
                        Bicompound_O.append(com)
```

```python
import re

original_combinations = ["M1N1", "M2N1", "M3N2","M4N1" "M1S1","M1S2","M2S3","M3S4","M6S8"]

compound_CNS = []  
index_CNS = []  
for original_combination in original_combinations:
    for replacement in metal:
        modified_combination = original_combination.replace("M", replacement, 1)

        matches = re.findall(r'(\d+)', modified_combination)

        if matches:
            indices = tuple(map(int, matches)) 

        index_CNS.append(indices)
        modified_combination_without_ones = re.sub(r'1(?!\d)', '', modified_combination)
        compound_CNS.append(modified_combination_without_ones)
```

```python
import re
import json

def parse_compound(compound):
    matches = re.findall(r'([A-Z][a-z]*)(\d*\.*\d*)', compound)
    
    result = {}
    
    for match in matches:
        element, count = match
        count = float(count) if count else 1.0  
        result[element] = count
    
    return result

def get_res(compound_name):
    parsed_result = parse_compound(compound_name)
    return parsed_result
```

```python
Bicompound_CNS = []
for ratio in element_ratio:
    for compound in compound_CNS:
        com = get_res(compound)
        key = list(com.keys())
        val = list(com.values())
        for za in second:
            if key[0] != za:
                za_index = round(ratio * val[0], 2)
                single_index = round((1 - ratio) * val[0], 2)
                tail_index = int(val[1])

                za_index_str = f"{za_index:.2f}".rstrip('0').rstrip('.') if za_index != 1.0 else ''
                single_index_str = f"{single_index:.2f}".rstrip('0').rstrip('.') if single_index != 1.0 else ''
                tail_index_str = str(tail_index) if tail_index != 1 else ''

                new_com = f"{za}{za_index_str}{key[0]}{single_index_str}{key[1]}{tail_index_str}"
                Bicompound_CNS.append(new_com)
Bicompound_CNS
```

```python
Tricompound_O = []

for compound in Bicompound_O:
    parsed = get_res(compound)  
    metals = [elem for elem in parsed.keys() if elem in metal] 

    if len(metals) != 2:
        continue  
    
    base_metal_1, base_metal_2 = metals
    base_metal_1_ratio = parsed[base_metal_1]
    base_metal_2_ratio = parsed[base_metal_2]
    oxygen_ratio = parsed["O"]

    for new_metal in second:
        if new_metal in metals:
            continue 

        for er in element_ratio: 
            new_metal_ratio = er * (base_metal_1_ratio + base_metal_2_ratio)
            
            updated_metal_1_ratio = base_metal_1_ratio - new_metal_ratio / 2
            updated_metal_2_ratio = base_metal_2_ratio - new_metal_ratio / 2

            total_metal_ratio = new_metal_ratio + updated_metal_1_ratio + updated_metal_2_ratio
            correction_factor = 1.0 / total_metal_ratio

            new_metal_ratio *= correction_factor
            updated_metal_1_ratio *= correction_factor
            updated_metal_2_ratio *= correction_factor

            oxygen_ratio_str = f"{oxygen_ratio:.2f}".rstrip('0').rstrip('.') if oxygen_ratio != 1 else ""

            new_compound = (
                f"{new_metal}{round(new_metal_ratio, 2) if round(new_metal_ratio, 2) != 1 else ''}"
                f"{base_metal_1}{round(updated_metal_1_ratio, 2) if round(updated_metal_1_ratio, 2) != 1 else ''}"
                f"{base_metal_2}{round(updated_metal_2_ratio, 2) if round(updated_metal_2_ratio, 2) != 1 else ''}"
                f"O{oxygen_ratio_str}"
            )

            if new_compound not in Tricompound_O:
                Tricompound_O.append(new_compound)

nums_triple = len(Tricompound_O)
print("nums_triple:", nums_triple)
```
