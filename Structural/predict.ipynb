# label

```python
import pandas as pd
import numpy as np
import joblib
import os

scaler = joblib.load('model/feature_label_scaler.pkl')

df = pd.read_csv('data/compound_new_TMCs_0506.csv')

feature_cols = [f'feat_{i}' for i in range(1, 26)]  # 1..26
missing = [c for c in feature_cols if c not in df.columns]
if missing:
    raise KeyError(f"no feature：{missing}")

X_new = df[feature_cols].values

X_new_scaled = scaler.transform(X_new)
```

```python

top_idx = joblib.load('model/label_top_idx.pkl')

X_new_selected = X_new_scaled[:, top_idx]  # (N, 10)


model_path = 'model/xgb_classifier.pkl'

clf = joblib.load(model_path)


threshold = 0.7  
y_proba_new = clf.predict_proba(X_new_selected)[:, 1]
y_pred_new = (y_proba_new >= threshold).astype(int)


compound_col = 'compound' if 'compound' in df.columns else (
    'compounds' if 'compounds' in df.columns else None
)
if compound_col is None:
    comp_series = pd.Series([None] * len(df), name='compound')
else:
    comp_series = df[compound_col].rename('compound')

save_df = pd.DataFrame({
    'compound': comp_series,
    'label': y_pred_new.astype(int)
})

os.makedirs('predict', exist_ok=True)
save_df.to_csv('predict/predicted_label_results.csv', index=False)
```

# Grain1

```python
import os, pandas as pd, joblib

os.makedirs('predict', exist_ok=True)

selected_features = joblib.load('model/grain1_selected_features.pkl')
model = joblib.load('model/grain1_xgb_model.pkl')

new_path = 'data/compound_new_TMCs_01.csv'
df_new = pd.read_csv(new_path)


for c in selected_features:
    if c not in df_new.columns:
        df_new[c] = 0

X_new = df_new[selected_features]

y_new = model.predict(X_new)
out = pd.DataFrame({
    'compound': df_new['compound'],
    'Grain_1': y_new
})

out.to_csv('predict/predicted_grain1_results.csv', index=False)
print('[Saved] predict/predicted_grain1_results.csv')
```

# grain_2

```python
import os
import pandas as pd, numpy as np, joblib

os.makedirs('predict', exist_ok=True)

pipe = joblib.load('model/grain2_pipeline.pkl')
df_new = pd.read_csv('data/compound_new_TMCs_1.csv')

X_new_full = df_new.drop(columns=[c for c in ['compound'] if c in df_new.columns],
                         errors='ignore').iloc[:, :25].values

sel = pipe.named_steps['select']
reg = pipe.named_steps['reg']

sel.fit(X_new_full)                      
X_new_sel = sel.transform(X_new_full)   
y_new = reg.predict(X_new_sel)

out = pd.DataFrame({
    'compound': df_new['compound'] if 'compound' in df_new.columns else np.arange(len(y_new)),
    'Grain_2': y_new
})

save_path = 'predict/predicted_grain2_results.csv'
out.to_csv(save_path, index=False)
print(f'[Saved] {save_path}  | rows={len(out)}  | cols={out.shape[1]}')
```
